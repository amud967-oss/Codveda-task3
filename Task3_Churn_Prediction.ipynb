```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Telco_Cleaned.csv')
df
                 
```

```python
sns.countplot(x='Contract', data=df)
plt.title('Churn by Contract')
plt.xlabel('Contract Type')
plt.ylabel('Count')
plt.savefig('bar_plot.png')
plt.show()
```

```python
df.groupby('tenure')['MonthlyCharges'].mean().plot()
plt.title('Line: Tenure vs MonthlyCharges')
plt.xlabel('Tenure')
plt.ylabel('MonthlyCharges')
plt.savefig('line_plot.png')
plt.show()

sns.scatterplot(x='tenure', y='TotalCharges', data=df)
plt.title('Scatter: Tenure vs TotalCharges')
plt.xlabel('Tenure')
plt.ylabel('TotalCharges')
plt.savefig('scatter_plot.png')
plt.show()
```
